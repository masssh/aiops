# Hello Agent - Interactive Notebook

このnotebookでは、hello_agentをインタラクティブに実行できます。

## 機能
- 異なるLLMプロバイダー（Gemini, OpenAI, Anthropic, Ollama）をテスト
- カスタムプロンプトで実行

```python
# セットアップ
import sys
sys.path.append('..')

from src.agents.hello_agent import hello_agent
from src.models.state import OverallState
from langchain_core.runnables import RunnableConfig
from dotenv import load_dotenv

load_dotenv()
```

```python
# 設定: LLMプロバイダーとモデルを選択
PROVIDER = "ollama"  # "gemini", "openai", "anthropic", "ollama" から選択
MODEL = "qwen3:8b"   # プロバイダーごとのモデル名（Noneでデフォルト使用）

# 初期状態
state: OverallState = {
    "products_config": {},
    "results": {}
}

# 設定
config = RunnableConfig(
    configurable={
        "provider": PROVIDER,
        "model": MODEL
    }
)
```

```python
# エージェントを実行
result = hello_agent(state, config)
print("\n実行完了!")
```

## 異なるプロバイダーで試す

```python
# Geminiで実行
print("=== Gemini ===")
hello_agent(state, RunnableConfig(configurable={"provider": "gemini"}))
```

```python
# Ollamaで実行（異なるモデル）
print("=== Ollama (qwen2.5-coder:7b) ===")
hello_agent(state, RunnableConfig(configurable={"provider": "ollama", "model": "qwen2.5-coder:7b"}))
```
