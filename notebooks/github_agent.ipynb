# GitHub Agent - Interactive Notebook

このnotebookでは、github_agentをインタラクティブに実行できます。

## 機能
- リポジトリのクローン
- Git操作（add, commit, push, pull）
- ブランチ管理
- Pull Request作成・管理
- Issue管理
- Release管理

## 必要なツール
- `git` コマンド
- `gh` CLI (GitHub CLI)

```python
# セットアップ
import sys
import os
import yaml
sys.path.append('..')

from src.agents.github_agent import github_agent
from src.models.state import OverallState
from langchain_core.runnables import RunnableConfig
from dotenv import load_dotenv

load_dotenv()
```

```python
# products.yamlをプロジェクトルートから読み込む
products_yaml_path = "../products.yaml"

if os.path.exists(products_yaml_path):
    with open(products_yaml_path, 'r', encoding='utf-8') as f:
        products_config = yaml.safe_load(f)
    print(f"✅ products.yaml を読み込みました")
    print(f"リポジトリ数: {len(products_config.get('repositories', []))}")
    for repo in products_config.get('repositories', []):
        print(f"  - {repo.get('id')}: {repo.get('name')}")
else:
    print(f"⚠️ products.yaml が見つかりません: {products_yaml_path}")
    products_config = {"repositories": [], "products": []}

# 初期状態（products.yamlから読み込んだ設定を使用）
state: OverallState = {
    "products_config": products_config,
    "results": {}
}
```

```python
# 設定: LLMプロバイダーとモデルを選択
PROVIDER = "ollama"  # "gemini", "openai", "anthropic", "ollama" から選択
MODEL = "qwen3:8b"   # プロバイダーごとのモデル名（Noneでデフォルト使用）
MAX_ITERATIONS = 20  # エージェントの最大反復回数
```

```python
PROMPT = """
リポジトリ springboot-microservices をcloneして
"""

config = RunnableConfig(
    configurable={
        "provider": PROVIDER,
        "model": MODEL,
        "github_agent_prompt": PROMPT,
        "max_iterations": MAX_ITERATIONS
    }
)

result = github_agent(state, config)
print("\n=== 実行完了! ===")
```
